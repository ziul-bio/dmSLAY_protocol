```python
import pandas as pd
from Bio.Seq import Seq
```

```python
data = pd.read_table('../results/counts/merged_counts.txt')
data['peptide'] = data['Sequence'].apply(lambda x: "".join(Seq(x).translate(to_stop=True)))
data = data.groupby(["peptide"], as_index=False).sum().copy()
data
```

```python
data['row_sum'] = data.iloc[:, 2:].sum(axis=1)
data.sort_values(by='row_sum', ascending=False, inplace=True)
data['peptide_length'] = data['peptide'].apply(len)
data
```
