```python
#!/usr/bin/env python3
"""Generate the PG-1 SLAY amplicon figure (SVG). Edit the CONFIG block and run:
    python make_amplicon_fig.py            # writes PG-1_SLAY_amplicon.svg
Optional PNG: pip install cairosvg
"""

# ---------------- CONFIG ----------------
TITLE = "PG-1 SLAY amplicon (232 bp)"
SEQ = ("aatgATACGGCGACCACCGAGATCTACACTCTTTCCCTACACGACGCTCTTCCGATCTCTCCAGCTGCGGGTATCGGAGGAACC"
       "nnnNNNnnnNNNnnnNNNnnnNN" "NnnnNNNnnnNNNnnnNNNnnnNNNnnnNNN" "TAAgtcgacctgcaggcatgcaagcttggc"
       "AGATCGGAAGAGCACACGTCTGAACTCCAGTCACNNNNNNATCTCGTATGCCGTCTTCTGCTTG")
# 1-based inclusive coordinates
LEFT, RIGHT, VAR = (59, 84), (139, 168), (85, 138)
LEFT_LABEL  = "5' Anchor"
RIGHT_LABEL = "3' Anchor"
COLORS = {  # (solid, light)
    "left":  ("#2f7dd1", "#d6e8fb"),
    "right": ("#e07b00", "#fde6c8"),
    "var":   ("#888888", "#eeeeee"),
}
PER_ROW, CW, X0, ROW_H, TOP = 60, 11, 60, 62, 130   # layout
OUT = "PG-1_SLAY_amplicon.svg"
MAKE_PNG = False
# ----------------------------------------

s = SEQ.upper()
n = len(s)
rows = (n + PER_ROW - 1) // PER_ROW
W_TOT = X0 + PER_ROW * CW + 30
H = TOP + rows * ROW_H + 60
SANS = "font-family='sans-serif'"

o = [f'<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 {W_TOT} {H}" '
     f'font-family="ui-monospace,Menlo,Consolas,monospace">',
     '<rect width="100%" height="100%" fill="#fff"/>',
     f'<text x="20" y="28" font-size="16" font-weight="bold" {SANS}>{TITLE}</text>']

# feature bar
bx, bw = 20, X0 + PER_ROW * CW - 10
sc = bw / n

def seg(a, b, fill, label, tc="#333"):
    x, w = bx + (a - 1) * sc, (b - a + 1) * sc
    o.append(f'<rect x="{x:.1f}" y="48" width="{w:.1f}" height="28" fill="{fill}" stroke="#555"/>')
    o.append(f'<text x="{x + w/2:.1f}" y="67" font-size="11" text-anchor="middle" fill="{tc}" {SANS}>{label}</text>')

seg(1, LEFT[0] - 1, "#ddd", "Illumina adapter (P5 / R1)")
seg(*LEFT, COLORS["left"][0], "left", "#fff")
seg(*VAR, "#fff", "variable CDS (N)")
seg(*RIGHT, COLORS["right"][0], "right", "#fff")
seg(RIGHT[1] + 1, n, "#ddd", "adapter (R2 / P7)")

for (a, b), key, lab in [(LEFT, "left", LEFT_LABEL), (RIGHT, "right", RIGHT_LABEL)]:
    x = bx + ((a + b) / 2 - 1) * sc
    o.append(f'<text x="{x:.1f}" y="96" font-size="12" text-anchor="middle" fill="{COLORS[key][0]}" '
             f'{SANS} font-weight="bold">{lab} {a}–{b}</text>')

# sequence rows
for r in range(rows):
    a = r * PER_ROW + 1
    chunk = s[r * PER_ROW:(r + 1) * PER_ROW]
    y = TOP + r * ROW_H
    o.append(f'<text x="20" y="{y+14}" font-size="11" fill="#888" {SANS}>{a}</text>')
    for i, ch in enumerate(chunk):
        p, x = a + i, X0 + i * CW
        fill = (COLORS["left"][1] if LEFT[0] <= p <= LEFT[1] else
                COLORS["right"][1] if RIGHT[0] <= p <= RIGHT[1] else
                COLORS["var"][1] if VAR[0] <= p <= VAR[1] else None)
        if fill:
            o.append(f'<rect x="{x}" y="{y}" width="{CW}" height="20" fill="{fill}"/>')
        o.append(f'<text x="{x + CW/2}" y="{y+15}" font-size="13" text-anchor="middle" fill="#222">{ch}</text>')
    for (A, B), key in [(LEFT, "left"), (RIGHT, "right")]:
        lo, hi = max(A, a), min(B, a + len(chunk) - 1)
        if lo <= hi:
            o.append(f'<rect x="{X0 + (lo-a)*CW}" y="{y+22}" width="{(hi-lo+1)*CW}" height="3" fill="{COLORS[key][0]}"/>')
    for p in range(a, a + len(chunk)):
        if p % 10 == 0:
            o.append(f'<text x="{X0 + (p-a)*CW + CW/2}" y="{y+38}" font-size="9" text-anchor="middle" fill="#999" {SANS}>{p}</text>')

# legend
ly = H - 24
for x, key, t in [(20, "left", "5' conserved region"),
                  (340, "right", "3' conserved region")]:
    o.append(f'<rect x="{x}" y="{ly-10}" width="14" height="14" fill="{COLORS[key][1]}" stroke="{COLORS[key][0]}"/>'
             f'<text x="{x+20}" y="{ly+2}" font-size="12" {SANS}>{t}</text>')
o.append("</svg>")

open(OUT, "w").write("\n".join(o))
print(f"wrote {OUT} ({n} bp)")
if MAKE_PNG:
    import cairosvg
    cairosvg.svg2png(url=OUT, write_to=OUT.replace(".svg", ".png"), scale=2.5)
```
